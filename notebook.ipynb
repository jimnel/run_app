```python
import calc
import re
```

```python
def from_str(s):
    assert bool(re.fullmatch(r"[0-9 ]+", s))
    x = s.split()
    assert len(x) <= 3

    if len(x) == 1:
        hours = int(x)
        mins = 0
    elif len(x) == 2:
            hours, mins = int(x)


    return calc.Time(hours, mins, 0)

from_str("1 30")
```
